```python
import seaborn as sns
import matplotlib.pyplot as plt

# Check counts numerically
print("Number of samples by day:")
print(df["Date"].value_counts())

# Plot the counts
plt.figure(figsize=(6,4))
sns.countplot(x="Date", data=df, palette="Set2")
plt.title("Number of Data Points per Storage Day")
plt.xlabel("Day")
plt.ylabel("Count of Samples")
plt.grid(axis="y", linestyle="--", alpha=0.6)
plt.show()
```

 each bar is roughly equal → your dataset is balanced, so day‑by‑day comparisons and classification modeling won’t be biased by sample size.
